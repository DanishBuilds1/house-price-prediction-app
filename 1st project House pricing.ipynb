```python
x = df.drop('Price',axis=1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print('x_train',x_train.shape)
print('x_test',x_test.shape)
print('y_train',y_train.shape)
print('y_test',y_test.shape)
```

```python
x_train.dtypes
```

```python
numeric_cols = x_train.select_dtypes(
    include=['int64', 'float64', 'int32']
).columns.tolist()

categorical_cols = x_train.select_dtypes(
    include=['category','object']
).columns.tolist()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

ctf = ColumnTransformer(
    transformers=[
        ('num',StandardScaler(),numeric_cols),
        ('cat',OneHotEncoder(handle_unknown='ignore'),categorical_cols)
        
    ]
)
```

```python
x_train_transformed = ctf.fit_transform(x_train)
x_test_transformed = ctf.transform(x_test)
```

```python
print("original x_train",x_train.shape)
print("transformed x_train",x_train_transformed.shape)
print("original x_test",x_test.shape)
print("transformed x_test",x_test_transformed.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train_transformed,y_train)
y_pred = lr.predict(x_test_transformed)
```

```python
print(y_pred[:10])
print(y_test.values[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_pred,y_test)
```

```python
alphas = [0.01, 0.1, 1, 10, 100]
```

```python
results = []

y_train_log = np.log1p(y_train)

for alpha in alphas:

    ridge = Ridge(alpha=alpha)

    ridge.fit(x_train_transformed, y_train_log)

    y_pred_log = ridge.predict(x_test_transformed)
    y_pred = np.expm1(y_pred_log)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    results.append([alpha, mae, mse, r2])

results_df = pd.DataFrame(
    results,
    columns=['alpha', 'MAE', 'MSE', 'R2']
)

results_df
```

```python
categorical_groups = {
    'Condition': [i for i, f in enumerate(feature_names)
                  if 'condition of the house_' in f],

    'Grade': [i for i, f in enumerate(feature_names)
              if 'grade of the house_' in f],

    'Waterfront': [i for i, f in enumerate(feature_names)
                   if 'waterfront present_' in f],

    'Postal Code': [i for i, f in enumerate(feature_names)
                    if 'Postal Code_' in f]
}

group_shap = {}

for group, indices in categorical_groups.items():
    group_shap[group] = abs(shap_values[:, indices]).sum(axis=1).mean()

group_shap_df = pd.DataFrame(
    list(group_shap.items()),
    columns=['Feature Group', 'Mean Absolute SHAP']
).sort_values(
    'Mean Absolute SHAP',
    ascending=False
)

group_shap_df
```

```python
group_shap = {}

for group, indices in categorical_groups.items():
    group_contribution = shap_values[:, indices].sum(axis=1)
    group_shap[group] = abs(group_contribution).mean()

group_shap_df = pd.DataFrame(
    list(group_shap.items()),
    columns=['Feature Group', 'Mean Absolute Group SHAP']
).sort_values(
    'Mean Absolute Group SHAP',
    ascending=False
)

group_shap_df
```

```python
# Create grouped SHAP importance for the original features

grouped_shap = {}

# Numerical features: one transformed feature per original feature
for col in num_cols:
    feature = f'num__{col}'
    indices = [i for i, f in enumerate(feature_names) if f == feature]

    if indices:
        grouped_shap[col] = abs(shap_values[:, indices[0]]).mean()

# Categorical features: combine all one-hot columns belonging to the original feature
for col in cat_cols:
    prefix = f'cat__{col}_'
    indices = [i for i, f in enumerate(feature_names) if f.startswith(prefix)]

    if indices:
        group_contribution = shap_values[:, indices].sum(axis=1)
        grouped_shap[col] = abs(group_contribution).mean()

grouped_shap_df = pd.DataFrame(
    list(grouped_shap.items()),
    columns=['Original Feature', 'Mean Absolute SHAP']
).sort_values(
    'Mean Absolute SHAP',
    ascending=False
)

grouped_shap_df
```

```python
def predict_house(raw_house):
    """
    Predict house price from the original 20 non-engineered features.

    Engineered features are created automatically.
    """

    house_df = pd.DataFrame([raw_house])

    # Training data is from 2016
    reference_year = 2016

    # Create engineered features
    house_df['Month'] = house_df['Sale Month']

    house_df['renovated'] = (
        house_df['Renovation Year'] != 0
    ).astype(int)

    house_df['House Age'] = (
        reference_year - house_df['Built Year']
    )

    house_df['year since renovation'] = np.where(
        house_df['Renovation Year'] == 0,
        0,
        reference_year - house_df['Renovation Year']
    )

    # Remove helper column
    house_df.drop(columns=['Sale Month'], inplace=True)

    predicted_log_price = loaded_pipeline.predict(house_df)

    predicted_price = np.expm1(predicted_log_price[0])

    return predicted_price
```
